# Netflix Content Exploratory Data Analysis

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("netflix1.csv")
```

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(
    x="type", data=df, palette=["cornflowerblue", "lightcoral"]
)
plt.title("Distribution of Netflix Movies & TV Shows", fontsize=14, fontweight="bold")
plt.xlabel("Type", fontsize=12)
plt.ylabel("Count", fontsize=12)
for p in ax.patches:
    ax.annotate(
        f"{int(p.get_height())}",
        (p.get_x() + p.get_width() / 2.0, p.get_height() / 2),
        ha="center",
        va="center",
        color="white",
        fontweight="bold",
        fontsize=12,
    )
plt.show()
```

```python
plt.figure(figsize=(12, 6))
year_counts = (
    df[df["release_year"] >= 2000]["release_year"].value_counts().sort_index()
)
sns.lineplot(
    x=year_counts.index,
    y=year_counts.values,
    marker="o",
    color="darkblue",
    linewidth=2.5,
)
plt.title(
    "Trend of Netflix Releases Over the Years (2000+)",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Number of Releases", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
top_countries = df["country"].value_counts().head(10)
sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    palette="viridis",
    hue=top_countries.index,
    legend=False,
)
plt.title(
    "Top 10 Countries Producing Netflix Content", fontsize=14, fontweight="bold"
)
plt.xlabel("Number of Titles", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.show()
```
